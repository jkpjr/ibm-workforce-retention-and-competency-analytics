### Load Pandas and the CSV

```python
import pandas as pd

df = pd.read_csv("WA_Fn-UseC_-HR-Employee-Attrition.csv")
```

### Flag Outliers

```python
# Define numerical columns with more than 10 unique values
for col in [col for col in df.select_dtypes(include='number') if df[col].nunique() > 10]:
    # Define IQR
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1

    # Define Upper and Lower limis
    ul = q3 + iqr * 1.5
    ll = q1 - iqr * 1.5

    # Create a new column to flag columns that has outliers
    if any(df[col] > ul) | any(df[col] < ll):
        df[f"{col}_is_outlier"] = ((df[col] > ul) | (df[col] < ll))
```

```python
df.to_csv("IBM-HR-Employee-Attrition-Outlier-Flagged.csv")
```
